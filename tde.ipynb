# Temporal Dictionary Ensemble (TDE)

Usage example for the TDE classifier from the paper "The Temporal Dictionary Ensemble (TDE) Classifier for Time Series Classification". See the [README](README.md) for the results from the paper.

TDE is available in the [aeon](https://github.com/aeon-toolkit/aeon) toolkit. It follows the `scikit-learn` interface and handles univariate and multivariate series.

```python
import warnings

from aeon.classification.dictionary_based import TemporalDictionaryEnsemble
from aeon.datasets import load_basic_motions, load_gunpoint
from sklearn.metrics import accuracy_score

warnings.filterwarnings("ignore")
```

## Univariate data

GunPoint is one of the 112 UCR datasets used in the paper.

```python
X_train, y_train = load_gunpoint(split="train")
X_test, y_test = load_gunpoint(split="test")

tde = TemporalDictionaryEnsemble(random_state=0)
tde.fit(X_train, y_train)

preds = tde.predict(X_test)
print(f"Accuracy: {accuracy_score(y_test, preds):.4f}")
```

## Multivariate data

TDE also handles multivariate series, such as the BasicMotions dataset.

```python
X_train, y_train = load_basic_motions(split="train")
X_test, y_test = load_basic_motions(split="test")

tde = TemporalDictionaryEnsemble(random_state=0)
tde.fit(X_train, y_train)

preds = tde.predict(X_test)
print(f"Accuracy: {accuracy_score(y_test, preds):.4f}")
```
